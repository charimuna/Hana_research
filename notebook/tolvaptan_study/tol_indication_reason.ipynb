```python

"""Hana Research: tolvaptan 導入理由候補の抽出（Step 4）

ind_date（サムスカ導入日）の30日前〜14日後のカルテから、
導入理由候補に関する記載をキーワード検索し、候補カルテをCSVに保存する。

注意：このスクリプトは候補カルテを抽出するStep 4のみを実装する。
患者単位の0/1表を作成するStep 5は、添付されたコードに含まれていない。
"""

import re
import sqlite3
import unicodedata
from pathlib import Path

import pandas as pd

DB_PATH = Path("/Users/muna/Hana_research/data/db/Hana_Research.db")
OUTPUT_DIR = Path("/Users/muna/Hana_research/src/tolvaptan_study/")
REASON_CANDIDATE_FILE = OUTPUT_DIR / "step4_reason_candidates.csv"

# 導入日前30日〜導入後14日
DAYS_BEFORE = 30
DAYS_AFTER = 14
NEGATION_WINDOW = 10

# 否定表現。NFKC正規化後は全角括弧などが半角に統一される。
NEGATION_REGEX = re.compile(
    r"(なし|無し|なく|無く|認めず|認めない|否定|ない|\(\s*[-－]\s*\))"
)

# 条件・指示文に見られる表現。キーワード直後の短い範囲だけを判定する。
CONDITION_REGEX = re.compile(
    r"(時|場合|あれば|ようなら|際|連絡|ください|下さい|有無|確認|チェック|リスク|注意|既往|疑い|可能性|恐れ|おそれ)"
)
# 読点・句点・改行をまたがず、最大10文字を許容
GAP = r"[^、。,;；\r\n]{0,10}"
WORSEN = r"(?:出現|増悪|悪化|増強|強まっ|強くな|出た|新規|増加)"

REASON_PATTERNS = {
    "weight_gain": [
        r"体重増加",
        r"体重.{0,10}増加",
        r"体重.{0,10}増",
        r"体重上昇",
        r"WEIGHT GAIN",
        r"\bBW.{0,10}増加",
        r"\bBW\s*(?:UP|↑)",
    ],
    "edema": [
        r"浮腫",
        r"むくみ",
        r"下腿浮腫",
        r"EDEMA",
    ],
    "dyspnea": [
        rf"(?:呼吸苦|呼吸困難|息切れ|息苦し[いさ]?|起座呼吸){GAP}{WORSEN}",
        rf"労作時呼吸困難{GAP}{WORSEN}",
        rf"(?<![A-Z])DOE(?![A-Z]){GAP}{WORSEN}",
        rf"(?:DYSPNEA|ORTHOPNEA){GAP}(?:WORSEN|INCREAS|ONSET|NEW)",
        rf"(?:新たに|新規){GAP}(?:呼吸苦|呼吸困難|息切れ|息苦し|起座呼吸)",
        r"(?:WORSENING|NEW\s+ONSET|NEW)\s+(?:DYSPNEA|ORTHOPNEA)",
    ],
    "fluid_retention": [
        r"水分貯留",
        r"体液貯留",
        r"うっ血",
        r"肺うっ血",
        r"CONGESTION",
    ],
    "diuretic_resistance": [
        r"利尿(?:剤|薬)?(?:抵抗|反応不良|効果不十分|効果不良|不良|無効)",
        r"REFRACTORY",
        r"ラシックス.{0,4}増量",
        r"ラシックス.{0,4}漸増",
        r"フロセミド.{0,4}増量",
        r"フロセミド.{0,4}漸増",
        r"ループ利尿薬.{0,4}増量",
        r"ループ利尿剤.{0,4}増量",
        r"利尿薬.{0,2}IV化",
        r"利尿剤.{0,2}IV化",
        r"利尿薬.{0,2}静注化",
        r"利尿剤.{0,2}静注化",
    ],
    "hyponatremia": [
        r"低Na血症",
        r"低ナトリウム血症",
        r"HYPONATREMIA",
    ],
    "pleural_effusion": [
        r"胸水",
        r"PLEURAL EFFUSION",
    ],
    "ascites": [
        r"腹水",
        r"ASCITES",
    ],
}

def has_unnegated_match(text: str, combined_regex: re.Pattern, negation_window: int = NEGATION_WINDOW) -> bool:
    """一致箇所の直後に否定・条件表現がない場合にTrueを返す。"""
    for match in combined_regex.finditer(text):
        # 検索語そのものに否定語が含まれる場合も除外
        if NEGATION_REGEX.search(match.group(0)):
            continue

        tail = text[match.end(): match.end() + negation_window]
        # 句読点・改行より後は、同じ記載の直後とはみなさない
        tail = re.split(r"[、。,;；\r\n]", tail)[0]

        if NEGATION_REGEX.search(tail):
            continue
        if CONDITION_REGEX.search(tail):
            continue
        return True

    return False


def run_step4() -> None:
    print("#" * 70)
    print(f"STEP 4: TOLVAPTAN REASON CANDIDATES (-{DAYS_BEFORE}d / +{DAYS_AFTER}d)")
    print("#" * 70)

    if not DB_PATH.is_file():
        raise FileNotFoundError(f"データベースが見つかりません: {DB_PATH}")
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    with sqlite3.connect(DB_PATH) as conn:
        tolvaptan = pd.read_sql_query(
            """
            SELECT Patient_ID, Study_ID, ind_date
            FROM tolvaptan_study
            WHERE ind_date IS NOT NULL
            """,
            conn,
        )

        if tolvaptan.empty:
            raise ValueError("tolvaptan_study に ind_date が登録された対象患者がいません。")

        tolvaptan["Patient_ID"] = tolvaptan["Patient_ID"].astype(str)
        tolvaptan["ind_date"] = pd.to_datetime(tolvaptan["ind_date"], errors="coerce")
        tolvaptan = tolvaptan.dropna(subset=["ind_date"])

        patient_ids = tolvaptan["Patient_ID"].dropna().unique().tolist()
        if not patient_ids:
            raise ValueError("有効な Patient_ID がありません。")

        placeholders = ",".join(["?"] * len(patient_ids))
        karte = pd.read_sql_query(
            f"""
            SELECT Patient_ID, Study_ID, visit_datetime, karte_text
            FROM karte
            WHERE Patient_ID IN ({placeholders})
              AND karte_text IS NOT NULL
              AND karte_text != ''
            """,
            conn,
            params=patient_ids,
        )

    karte["Patient_ID"] = karte["Patient_ID"].astype(str)
    karte["visit_date"] = pd.to_datetime(
        karte["visit_datetime"].astype(str).str.extract(r"^(\d{4}/\d{1,2}/\d{1,2})", expand=False),
        errors="coerce",
    )

    # 患者IDで導入日を付与
    tolvaptan_merge = tolvaptan[["Patient_ID", "Study_ID", "ind_date"]].rename(
        columns={"Study_ID": "Study_ID_tolva"}
    )
    karte = karte.merge(tolvaptan_merge, on="Patient_ID", how="inner")
    karte["days_from_ind"] = (karte["visit_date"] - karte["ind_date"]).dt.days

    karte_window = karte[
        karte["visit_date"].notna()
        & karte["ind_date"].notna()
        & (karte["days_from_ind"] >= -DAYS_BEFORE)
        & (karte["days_from_ind"] <= DAYS_AFTER)
    ].copy()

    print(f"対象患者数: {len(tolvaptan):,}")
    print(f"取得カルテ件数: {len(karte):,}")
    print(f"導入日前{DAYS_BEFORE}日〜導入後{DAYS_AFTER}日のカルテ件数: {len(karte_window):,}")

    karte_window["karte_text_search"] = karte_window["karte_text"].astype(str).map(
        lambda s: unicodedata.normalize("NFKC", s)
    ).str.upper()

    hits = []
    for reason, patterns in REASON_PATTERNS.items():
        regex = re.compile("|".join(patterns), flags=re.IGNORECASE)
        mask = karte_window["karte_text_search"].apply(
            lambda text: has_unnegated_match(text, regex)
        )
        tmp = karte_window.loc[
            mask,
            [
                      "Study_ID_tolva",
                "Patient_ID",
                "ind_date",
                "visit_date",
                "days_from_ind",
                "visit_datetime",
                "karte_text",
            ],
        ].copy()
        tmp["reason"] = reason
        hits.append(tmp)

   # 該当ゼロの項目があっても、出力列を維持して保存できるようにする
    if hits:
        result = pd.concat(hits, ignore_index=True)
    else:
        result = pd.DataFrame()

    if result.empty:
        result = pd.DataFrame(
            columns=[
                "Study_ID_tolva", "Patient_ID", "ind_date", "visit_date",
                "days_from_ind", "visit_datetime", "karte_text", "reason"
            ]
        )

    result = result.rename(columns={"Study_ID_tolva": "Study_ID"})
    result = result[
        [
            "Study_ID", "Patient_ID", "ind_date", "reason", "visit_date",
            "days_from_ind", "visit_datetime", "karte_text"
        ]
    ]
    result.to_csv(REASON_CANDIDATE_FILE, index=False, encoding="utf-8-sig")

    print(f"候補カルテCSV: {REASON_CANDIDATE_FILE}")
    print("理由別ヒット件数:")
    if result.empty:
        print("該当なし")
    else:
        print(result["reason"].value_counts().reindex(REASON_PATTERNS.keys(), fill_value=0))


if __name__ == "__main__":
    run_step4()
```
