```python
# ============================================
# サムスカ／トルバプタン新規導入候補抽出
# ============================================

import sqlite3
import pandas as pd
import re

DB_PATH = "/Users/muna/Hana_research/data/db/Hana_Research.db"
OUTPUT_CSV = "samsca_start_candidates.csv"

# --------------------------------------------
# 2025年以降で薬剤名を含むカルテを取得
# --------------------------------------------
conn = sqlite3.connect(DB_PATH)

query = """
SELECT
    record_no,
    Patient_ID,
    visit_datetime,
    karte_text
FROM karte
WHERE
(
    visit_datetime LIKE '2025/%'
    OR visit_datetime LIKE '2026/%'
)
AND
(
    karte_text LIKE '%サムスカ%'
    OR karte_text LIKE '%トルバプタン%'
)
"""

df = pd.read_sql(query, conn)

print(f"対象カルテ数: {len(df):,}")

# --------------------------------------------
# 開始候補キーワード
# --------------------------------------------
start_patterns = [
    r"開始",
    r"導入",
    r"追加",
    r"処方",
    r"開始した",
    r"開始する",
    r"開始とな",
    r"導入開始",
    r"導入する",
]

# --------------------------------------------
# 除外キーワード
# --------------------------------------------
exclude_patterns = [
    r"継続",
    r"維持",
    r"増量",
    r"減量",
    r"中止",
    r"開始予定",
    r"導入予定",
    r"開始検討",
    r"導入検討",
    r"希望せず",
    r"以前から内服",
]

# --------------------------------------------
# 前後100文字取得
# --------------------------------------------
def get_context(text, start, end, window=100):
    left = max(0, start - window)
    right = min(len(text), end + window)
    return text[left:right].replace("\n", " ")

# --------------------------------------------
# 判定
# --------------------------------------------
results = []

for _, row in df.iterrows():

    text = str(row["karte_text"])

    hit = False
    hit_keyword = None
    hit_context = None

    # サムスカまたはトルバプタン出現位置
    for drug_match in re.finditer(r"(サムスカ|トルバプタン)", text):

        drug_start = drug_match.start()

        # 前後80文字を見る
        local_start = max(0, drug_start - 80)
        local_end = min(len(text), drug_start + 80)

        nearby = text[local_start:local_end]

        # 除外語
        if any(re.search(p, nearby) for p in exclude_patterns):
            continue

        # 開始語
        matched_start = None

        for p in start_patterns:
            m = re.search(p, nearby)
            if m:
                matched_start = m.group()
                break

        if matched_start:

            hit = True
            hit_keyword = (
                drug_match.group() + "_" + matched_start
            )

            hit_context = get_context(
                text,
                drug_match.start(),
                drug_match.end(),
                100
            )

            break

    if hit:
        results.append({
            "record_no": row["record_no"],
            "Patient_ID": row["Patient_ID"],
            "visit_datetime": row["visit_datetime"],
            "hit_keyword": hit_keyword,
            "context": hit_context
        })

# --------------------------------------------
# 結果DataFrame作成
# --------------------------------------------
result_df = pd.DataFrame(results)

# --------------------------------------------
# tolvaptan_study に登録済みの Patient_ID を除外
# --------------------------------------------

study_df = pd.read_sql("""
SELECT DISTINCT Patient_ID
FROM tolvaptan_study
WHERE Patient_ID IS NOT NULL
""", conn)

study_ids = set(
    study_df["Patient_ID"]
    .astype(str)
    .str.strip()
)

before_n = len(result_df)

result_df = result_df[
    ~result_df["Patient_ID"].astype(str).str.strip().isin(study_ids)
].copy()

after_n = len(result_df)

print(f"除外前: {before_n:,}")
print(f"tolvaptan_study登録済み除外後: {after_n:,}")
print(f"除外数: {before_n-after_n:,}")

# --------------------------------------------
# Patient_IDごとに最古のvisit_datetimeのみ残す
# --------------------------------------------

# 元の日時文字列を保存
result_df["visit_datetime_original"] = result_df["visit_datetime"]

# "(木)"など曜日部分を除去
result_df["visit_datetime_dt"] = (
    result_df["visit_datetime"]
    .str.replace(r"\([^)]*\)", "", regex=True)
)

# datetime型へ変換
result_df["visit_datetime_dt"] = pd.to_datetime(
    result_df["visit_datetime_dt"],
    format="%Y/%m/%d %H:%M",
    errors="coerce"
)

# Patient_IDごとに最古日時順に並べる
result_df = result_df.sort_values(
    ["Patient_ID", "visit_datetime_dt"]
)

# Patient_IDごとに最初の1件だけ残す
before_dedup = len(result_df)

result_df = result_df.drop_duplicates(
    subset=["Patient_ID"],
    keep="first"
).copy()

after_dedup = len(result_df)

print(f"重複除去前: {before_dedup:,}")
print(f"Patient_IDごと最古1件: {after_dedup:,}")
print(f"削除行数: {before_dedup-after_dedup:,}")

# 補助列削除
result_df = result_df.drop(
    columns=["visit_datetime_dt", "visit_datetime_original"],
    errors="ignore"
)

# --------------------------------------------
# 保存
# --------------------------------------------
result_df = result_df.sort_values(
    ["visit_datetime", "Patient_ID"]
)

result_df.to_csv(
    OUTPUT_CSV,
    index=False,
    encoding="utf-8-sig"
)

print()
print(f"抽出件数: {len(result_df):,}")
print(f"保存先: {OUTPUT_CSV}")

display(result_df.head(20))
```
