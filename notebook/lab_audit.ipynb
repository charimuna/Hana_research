```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("/Users/muna/Hana_research/data/db/Hana_Research.db")

# lab_clean から CRE だけ確認
df_cre = pd.read_sql("""
    SELECT item_code, COUNT(*) as 件数
    FROM lab_clean
    WHERE item_name = 'CRE'
    GROUP BY item_code
""", conn)

print(df_cre)
print("\nitem_codeの型:", df_cre["item_code"].dtype)
```

```python
df_cre_names = pd.read_sql("""
    SELECT DISTINCT item_name, item_code, COUNT(*) as 件数
    FROM lab_clean
    WHERE item_name LIKE '%CRE%'
       OR item_name LIKE '%cre%'
       OR item_name LIKE '%Cre%'
       OR item_name LIKE '%クレ%'
    GROUP BY item_name, item_code
    ORDER BY item_name
""", conn)

print(df_cre_names.to_string())
```

```python
df_code_check = pd.read_sql("""
    SELECT item_name, item_code, COUNT(*) as 件数
    FROM lab_clean
    WHERE item_code IN ('0069', '0074')
    GROUP BY item_name, item_code
    ORDER BY item_code
""", conn)

print(df_code_check.to_string())
```

```python
# 同じitem_nameで複数item_codeが存在する項目を一覧表示
conn = sqlite3.connect("/Users/muna/Hana_research/data/db/Hana_Research.db")

df_dup_codes = pd.read_sql("""
    SELECT item_name, COUNT(DISTINCT item_code) as コード数,
           GROUP_CONCAT(DISTINCT item_code) as コード一覧
    FROM lab_clean
    GROUP BY item_name
    HAVING COUNT(DISTINCT item_code) > 1
    ORDER BY コード数 DESC
""", conn)

print(df_dup_codes.to_string())
```
