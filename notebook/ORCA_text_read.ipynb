```python
from pathlib import Path
import chardet

file_path = Path(
    "/Users/muna/Hana_research/data/raw/ORCAから処方/42704.txt"
)

with open(file_path, "rb") as f:
    raw = f.read()

print("ファイルサイズ:", len(raw), "bytes")
print("先頭32 bytes:", raw[:32])
print("末尾32 bytes:", raw[-32:])
print("文字コード推定:", chardet.detect(raw))

# エラーが発生した位置周辺を表示
error_position = 126094

start = max(0, error_position - 30)
end = min(len(raw), error_position + 30)

print("エラー位置周辺のバイト列:")
print(raw[start:end])
```
