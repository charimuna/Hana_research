```python
import sqlite3
import pandas as pd

db_path = '/Users/muna/Hana_research/data/db/Hana_Research.db'
conn = sqlite3.connect(db_path)

# サムスカ導入 かつ 心疾患の患者を抽出
query = """
SELECT bs.*
FROM Background_summary bs
INNER JOIN first_diag fd ON bs.patient_id = fd.patient_id
WHERE bs.memo LIKE '%サムスカ導入%'
  AND fd.category = '【心疾患】'
"""

df = pd.read_sql_query(query, conn)
conn.close()

print(f"該当患者数: {len(df)}件")
df
```

```python
conn = sqlite3.connect(db_path)

# first_diagで心疾患の患者ごとの行数を確認
check = pd.read_sql_query("""
    SELECT patient_id, COUNT(*) as count
    FROM first_diag
    WHERE category = '【心疾患】'
    GROUP BY patient_id
    HAVING COUNT(*) > 1
""", conn)

print(f"複数行ある患者: {len(check)}件")
print(check)
conn.close()
```

```python
conn = sqlite3.connect(db_path)

query = """
SELECT DISTINCT bs.*
FROM Background_summary bs
INNER JOIN first_diag fd ON bs.patient_id = fd.patient_id
WHERE bs.memo LIKE '%サムスカ導入%'
  AND fd.category = '【心疾患】'
"""

df = pd.read_sql_query(query, conn)
conn.close()

print(f"該当患者数: {len(df)}件")
df
```

```python
import sqlite3
import pandas as pd
import re

db_path = '/Users/muna/Hana_research/data/db/Hana_Research.db'
conn = sqlite3.connect(db_path)

df = pd.read_sql("SELECT Patient_ID, memo FROM Background_summary WHERE memo LIKE '%サムスカ導入%'", conn)
conn.close()

# 日付が抽出できなかった行を表示
def extract_date(memo):
    match = re.search(r'サムスカ導入\((\d{4})/(\d{1,2})/(\d{1,2})\)', memo)
    return match is not None

df["extracted"] = df["memo"].apply(extract_date)
no_match = df[~df["extracted"]]

print(f"日付抽出できなかった件数: {len(no_match)}")
for _, row in no_match.iterrows():
    # サムスカ導入前後30文字だけ表示
    idx = row["memo"].find("サムスカ導入")
    print(f"\nPatient_ID: {row['Patient_ID']}")
    print(row["memo"][max(0, idx-5):idx+30])
```
