```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("/Users/muna/Hana_research/data/db/Hana_Research.db")
cursor = conn.cursor()
print("接続OK")
```

```python
cursor.execute("SELECT name FROM sqlite_master WHERE type='table'")
tables = cursor.fetchall()
print("テーブル一覧:")
for t in tables:
    print(" -", t[0])
```

```python
tables = [
    "Patient_Master", "first_diag", "event", "intervention_history",
    "unexpected_death", "Background_summary", "unex_study",
    "Freedocument", "study_id_linkage", "lab_raw", "lab_clean"
]

for t in tables:
    cursor.execute(f"SELECT COUNT(*) FROM {t}")
    count = cursor.fetchone()[0]
    print(f"{t}: {count}件")
```

```python
for t in tables:
    cursor.execute(f"PRAGMA table_info({t})")
    cols = [row[1] for row in cursor.fetchall()]
    print(f"\n{t}")
    print(" ", cols)
```
